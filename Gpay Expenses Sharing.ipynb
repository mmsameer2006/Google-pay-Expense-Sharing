```python

class ExpenseSharing:

    def __init__(self):
        self.friends = ["Alice", "Bob", "Carol"]
        self.expenses = []

    # Add Expenses
    def add_expenses(self):

        print("\n--- Add Trip Expenses ---")

        n = int(input("Enter number of expenses: "))

        for i in range(n):

            print("\nExpense", i + 1)

            expense_type = input(
                "Enter expense type (Accommodation/Meals): "
            )

            name = input(
                "Who paid? (Alice/Bob/Carol): "
            )

            if name not in self.friends:
                print("Invalid friend name!")
                continue

            amount = float(input("Enter amount: "))

            if amount <= 0:
                print("Invalid amount!")
                continue

            self.expenses.append(
                [expense_type, name, amount]
            )

            print("Expense added successfully!")


    # Display Expenses
    def display_expenses(self):

        print("\n--- All Trip Expenses ---")

        for expense in self.expenses:

            print("Expense Type:", expense[0])
            print("Paid By:", expense[1])
            print("Amount: Rs.", expense[2])
            print("------------------------")


    # Calculate Expenses
    def calculate_expenses(self):

        total = 0

        paid = {
            "Alice": 0,
            "Bob": 0,
            "Carol": 0
        }

        for expense in self.expenses:

            name = expense[1]
            amount = expense[2]

            paid[name] = paid[name] + amount

            total = total + amount

        share = total / 3

        print("\n--- Expense Summary ---")

        print("Total Expense: Rs.", round(total, 2))

        print("Each Person Share: Rs.", round(share, 2))

        # Calculate Balance

        debtors = []
        creditors = []

        for name in self.friends:

            balance = round(paid[name] - share, 2)

            if balance < 0:
                debtors.append([name, -balance])

            elif balance > 0:
                creditors.append([name, balance])

        # Final Settlement

        print("\n--- Final Settlement ---")

        i = 0
        j = 0

        while i < len(debtors) and j < len(creditors):

            amount = min(
                debtors[i][1],
                creditors[j][1]
            )

            print(
                debtors[i][0],
                "should pay Rs.",
                round(amount, 2),
                "to",
                creditors[j][0]
            )

            debtors[i][1] = round(debtors[i][1] - amount, 2)

            creditors[j][1] = round(creditors[j][1] - amount, 2)

            if debtors[i][1] == 0:
                i += 1

            if creditors[j][1] == 0:
                j += 1


    # Main Function
    def main(self):

        self.add_expenses()

        self.display_expenses()

        self.calculate_expenses()


# Create Object
obj = ExpenseSharing()

# Run Project

obj.main()
```
